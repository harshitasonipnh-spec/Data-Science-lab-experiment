```python
# que21
sales = np.array([
    [120, 150, 180],
    [140, 160, 175],
    [135, 155, 190],
    [150, 170, 200]
])

daily_sales = np.sum(sales, axis=1)

product_sales = np.sum(sales, axis=0)

highest_day = np.argmax(daily_sales) + 1

print("Daily Sales:", daily_sales)
print("Product Sales:", product_sales)
print("Day with Highest Sales:", highest_day)
```

```python
# que22
a = np.arange(10, 46).reshape(6, 6)
print("Original Array:")
print(a)

sub = a[2:6, 1:5]
print("Extracted Submatrix:")
print(sub)

print("Mean:", np.mean(sub))
```

```python
# que23
weights = np.array([48, 62, 75, 81, 54, 69, 90])

above_70 = weights[weights > 70]
new_weights = np.where(weights < 55, 55, weights)

print("Weights above 70:", above_70)
print("New Weights:", new_weights)
print("Original Weights:", weights)
```

```python
# que24
quantity = np.array([3, 5, 2, 7])
price = np.array([120, 80, 250, 60])

total_cost = quantity * price
bill = np.sum(total_cost)
percentage = (total_cost / bill) * 100

print("Total Cost:", total_cost)
print("Overall Bill:", bill)
print("Percentage Contribution:", percentage)
```

```python
# que25
a = np.array([
    [10, 20, 30, 40],
    [15, 25, 35, 45],
    [20, 30, 40, 50]
])
print("Original Array:")
print(a)
print("Row Variance:", np.var(a, axis=1))
print("Row Standard Deviation:", np.std(a, axis=1))
print("Column Variance:", np.var(a, axis=0))
print("Column Standard Deviation:", np.std(a, axis=0))
```

```python
# que29
dept_a = pd.Series({'Jan': 250, 'Feb': 310, 'Mar': 290})
dept_b = pd.Series({'Mar': 180, 'Jan': 210, 'Apr': 260})
result = dept_a + dept_b
print("Department A:")
print(dept_a)
print("\nDepartment B:")
print(dept_b)
print("\nTotal Revenue:")
print(result)
print("\nLabels producing missing values:")
print(result[result.isnull()].index.tolist())
```

```python
# que30
attendance = pd.Series(
    [82, 67, 91, 74, 58],
    index=['S101', 'S102', 'S103', 'S104', 'S105']
)
below_75 = attendance[attendance < 75]
print("Students with attendance below 75%:")
print(below_75.index.tolist())
print("Total students:", len(below_75))
```

```python
# que31
inventory = {
    'Pen': 50,
    'Pencil': 80,
    'Eraser': 40,
    'Scale': 25
}
labels = ['Scale', 'Pen', 'Marker', 'Eraser']
s = pd.Series(inventory, index=labels)
print(s)
```

```python
# que32
s1 = pd.Series(
    [100, 200, 300],
    index=['North', 'South', 'East']
)
s2 = pd.Series(
    [10, 20, 30],
    index=['South', 'West', 'North']
)
result = s1 - s2
print("Subtraction Result:")
print(result)
print("\nNon-missing Entries:")
print(result[result.notnull()])
```

```python
# que33
import pandas as pd
electricity = pd.Series(
    [180, 280, 320, 220, 400],
    index=['House1', 'House2', 'House3', 'House4', 'House5']
)
print("Original Series:")
print(electricity)
filtered = electricity[electricity > 250]
print("\nConsumption above 250 units:")
print(filtered)
data = filtered.to_dict()
print("\nDictionary:")
print(data)
new_series = pd.Series(data)
print("\nNew Series:")
print(new_series)
```

```python
# que34
s = pd.Series(
    {'A': 45, 'B': 70, 'C': 55},
    index=['C', 'D', 'A', 'B']
)
s.name = 'Performance'
s.index.name = 'Group'
print("Filtered Series:")
print(s[s >= 50])
print("\nMissing Entries:")
print(s[s.isnull()])
print("\nSeries Name:", s.name)
print("Index Name:", s.index.name)
```
