```python
# que29
data = {
    'Product': ['Laptop', 'Mobile', 'Keyboard', 'Mouse', 'Monitor'],
    'Price': [50000, 20000, 1500, 800, 12000]
}
inventory_df = pd.DataFrame(data)
stock = pd.Series(
    [10, 20, 30],
    index=[0, 2, 4]
)
inventory_df['Stock'] = stock
print("DataFrame after adding Stock:")
print(inventory_df)
inventory_df = inventory_df.drop('Stock', axis=1)
print("\nDataFrame after deleting Stock:")
print(inventory_df)
```

```python

import pandas as pd
# Create DataFrame
data = {
    'Name': ['Rahul', 'Priya', 'Aman'],
    'Age': [20, 21, 19],
    'City': ['Delhi', 'Agra', 'Mathura'],
    'Total Marks': [85, 90, 78]
}
lab_df = pd.DataFrame(data)

print("Original DataFrame:")
print(lab_df)

print("\nColumn Names:")
print(lab_df.columns)

print("\nName Column:")
print(lab_df['Name'])

print("\nTotal Marks Column:")
print(lab_df['Total Marks'])

new_df = lab_df[['Name', 'Age', 'City']]
print("\nSmaller DataFrame:")
print(new_df)

lab_df['Grade'] = ['A', 'A+', 'B']
print("\nAfter Adding Grade:")
print(lab_df)

lab_df.drop('Grade', axis=1, inplace=True)
print("\nAfter Deleting Grade:")
print(lab_df)
```
