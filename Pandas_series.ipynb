```python
#que29
sales = pd.Series(
    [300, 650, 800, 450],
    index=['Pen', 'Bag', 'Laptop', 'Book']
)

# Display values
print("Values:")
print(sales.values)

# Display index
print("Index:")
print(sales.index)

# Select one label
print("Laptop sales:")
print(sales['Laptop'])

# Filter sales above 500
print("Sales above 500:")
print(sales[sales > 500])

# Check whether a label exists
print("Is Bag present?")
print('Bag' in sales.index)

# Convert Series to dictionary
sales_dict = sales.to_dict()
print("Dictionary:")
print(sales_dict)
```
